```python
df["AgeGroup"] = pd.cut(df["Age"], bins=[17, 29, 44, 59, 100], labels=["18-29", "30-44", "45-59", "60+"])
df["PolicyMonth"] = df["PolicyStartDate"].dt.strftime("%Y-%m")
```

```python
settled = df[df["ClaimStatus"] == "Settled"]
settled.groupby("PolicyType")["ClaimAmount"].sum()
settled.groupby("AgeGroup",observed=True)["ClaimAmount"].sum()
```
